<b> Description of the Sample Application:</b> Imagine a black box, or a mystery ML model, that takes two numbers as input and returns a log-likelihood score. Your goal is to maximise that score, but each output is noisy, and depending on where you start, you might get stuck in a local optimum. 
To tackle this, you use Bayesian optimisation, which selects the next inputs based on what it has learned so far. It balances exploration with exploitation, making it well suited to noisy outputs and complex functions with many local peaks. <br>

# Compute the UCB acquisition function

- UCB(x)=μ(x)+β⋅σ(x)
- A high beta favors exploration (uncertainty) over exploitation (predicted mean).
- **Week 1:** uses beta = 2.5, reflecting the 90:10 exploration-heavy strategy.
- **Week 2:** β is lowered to **2.0**, to shift from exploration toward exploitation.

```python
grid_res   = 500
xs         = np.linspace(0, 1, grid_res)
ys         = np.linspace(0, 1, grid_res)
gx, gy     = np.meshgrid(xs, ys)
candidates = np.column_stack([gx.ravel(), gy.ravel()])

mu, std    = gpr.predict(candidates, return_std = True)

# Setting a higher beta(= 2.5) for Week -1
#beta       = 2.5    #Week 1 
beta       = 2.0    #Week 2 
ucb        = mu + beta * std

best_ucb_idx = np.argmax(ucb)
next_point   = candidates[best_ucb_idx]

print(f"Suggested next Query Point: {next_point}")
print(f"Predicted Mean(μ): {mu[best_ucb_idx]:.6e}")
print(f"Predicted Std (σ): {std[best_ucb_idx]:.6e}")
print(f"UCB Score: {ucb[best_ucb_idx]:.6e}")
```

# Week 3: Consolidated Bayesian Optimization Function

```python
# Week 3: Consolidated Bayesian Optimization Function
def bayesian_optimization_step(X, y, bounds, beta = 2.0, kernel = None,
                                n_restarts = 20, random_state = 42):
    if kernel is None:
        kernel = (ConstantKernel(1.0, (1e-1, 1e2))
                  * RBF(length_scale = [0.25, 0.25], length_scale_bounds = (1e-2, 2.0))
                  + WhiteKernel(noise_level = 1e-2, noise_level_bounds = (1e-5, 1e0)))

    gp         = GaussianProcessRegressor(kernel=kernel, normalize_y = True,
                                  n_restarts_optimizer = 10, random_state = random_state)
    gp.fit(X, y)

    def neg_ucb(x):
        x         = np.array(x).reshape(1, -1)
        mu, sigma = gp.predict(x, return_std = True)
        return -(mu[0] + beta * sigma[0])

    lows          = [b[0] for b in bounds]
    highs         = [b[1] for b in bounds]

    best_val      = np.inf
    best_x        = None
    rng           = np.random.default_rng(random_state)
    for _ in range(n_restarts):
        x0        = rng.uniform(lows, highs)
        result    = minimize(neg_ucb, x0, bounds = bounds, method = 'L-BFGS-B')
        if result.fun < best_val:
            best_val = result.fun
            best_x   = result.x

    mu_next, std_next = gp.predict(best_x.reshape(1, -1), return_std = True)

    return {
        'gp': gp,
        'next_point': best_x,
        'mu': mu_next[0],
        'std': std_next[0],
        'ucb': -best_val,
    }
```

```python
# Run the consolidated function for Week 3
# lowered value of beta from Week 2's 2.0 to 1.5
beta             = 1.5   
bounds_2d        = [(0.0, 1.0), (0.0, 1.0)]

result_week3     = bayesian_optimization_step(X, Y, bounds_2d, beta = beta)

next_point_week3 = result_week3['next_point']

print(f"Suggested Week 3 Query Point: {next_point_week3}")
print(f"Predicted Mean(\u03bc): {result_week3['mu']:.6e}")
print(f"Predicted Std (\u03c3): {result_week3['std']:.6e}")
print(f"UCB Score: {result_week3['ucb']:.6e}")
```

# Week 4: Kernel Re-tuning Check
Across Weeks 1–3, the fitted RBF kernel's x2 length-scale has repeatedly settled at or near its upper bound (2.0), triggering a scikit-learn convergence warning each time. This can mean the model wants an even larger length-scale than allowed (i.e., it thinks the function is smoother/flatter along x2 than the bound permits). Worth checking directly rather than ignoring the warning again this week.

```python
# Original kernel (bounds used since Week 1): length_scale_bounds = (1e-2, 2.0)
log_ml_original  = gpr.log_marginal_likelihood(gpr.kernel_.theta)
print(f"Original kernel:              {gpr.kernel_}")
print(f"Log-marginal-likelihood:      {log_ml_original:.4f}")

# Widened-bounds candidate: let the length-scale optimiser search further,
# in case the true optimum lies beyond the original 2.0 cap
kernel_rbf_wide  = ConstantKernel(1.0, (1e-1, 1e2)) * RBF(length_scale = [0.25, 0.25], length_scale_bounds = (1e-2, 5.0)) \
                    + WhiteKernel(noise_level = 1e-2, noise_level_bounds = (1e-5, 1e0))
gpr_wide = GaussianProcessRegressor(kernel = kernel_rbf_wide, normalize_y = True,
                                     n_restarts_optimizer = 10, random_state = 42)
gpr_wide.fit(X, Y)
log_ml_wide      = gpr_wide.log_marginal_likelihood(gpr_wide.kernel_.theta)
print(f"\nWidened-bounds kernel:        {gpr_wide.kernel_}")
print(f"Log-marginal-likelihood:      {log_ml_wide:.4f}")

# Adopt whichever kernel fits better (higher log-marginal-likelihood)
if log_ml_wide > log_ml_original:
    print("\n-> Widened-bounds kernel fits better; using it for Week 4's acquisition step.")
    kernel_week4 = gpr_wide.kernel_
else:
    print("\n-> Original kernel bounds were already adequate; keeping them for Week 4.")
    kernel_week4 = gpr.kernel_
```

# Week 13 (Final): Surrogate Bake-off, Two Candidate Final Queries

```python
# Checking how good was the Week 12 prediction (using a frozen 21-point snapshot)
X21, Y21 = X_updated[:21], Y_updated[:21]
X22, Y22 = X_updated[:22], Y_updated[:22] 

kernel_w13_check = ConstantKernel(1.0, (1e-1, 1e2)) * RBF(length_scale=[0.25, 0.25], length_scale_bounds=(1e-2, 5.0)) \
                     + WhiteKernel(noise_level=1e-2, noise_level_bounds=(1e-5, 1e0))
gpr_pre_w12 = GaussianProcessRegressor(kernel=kernel_w13_check, normalize_y=True, n_restarts_optimizer=10, random_state=42)
gpr_pre_w12.fit(X21, Y21)
mu_p, sd_p = gpr_pre_w12.predict(X_w12_new_point.reshape(1, -1), return_std=True)
z_w12 = (Y_w12_new_point[0] - mu_p[0]) / sd_p[0]
print(f"Pre-Week-12 model predicted mu={mu_p[0]:.4f}, std={sd_p[0]:.4f} at {X_w12_new_point}")
print(f"Actual outcome: {Y_w12_new_point[0]:.4f}  ->  {z_w12:+.2f} predicted std devs from the prediction")

gpr_w13 = GaussianProcessRegressor(kernel=kernel_w13_check, normalize_y=True, n_restarts_optimizer=10, random_state=42)
gpr_w13.fit(X_updated, Y_updated)
print(f"\nKernel on all 22 points: {gpr_w13.kernel_}")
print("Current best (unchanged):", X_updated[np.argmax(Y_updated)], "-> y =", Y_updated.max())
```

**Observation:** After three consecutive misses, Week 12 is the **first query in four weeks that landed inside the model's 
uncertainty band** (−0.6σ), so the global model is not broken everywhere. **Correction to my own earlier reading:** I first took this singl 
point (0.7, 0.3 → 0.534) as showing the x2 gap is a plateau with no hidden peak. One observation cannot show that. The Week 9–11 clust r
already demonstrated that outputs in this band can change by 0.26 over an x2 step of 0.014, so a gap 0.25 wide with one point at its edge is
essentially **unexplored**, not 'ruled out'. Nothing has beaten 0.711209 yet.

```python
# Surrogate bake-off
# Leave-one-out CV on all 22 points for six model families. 'local band' = the x1 in (0.65, 0.76) points,
# i.e. the region the final query will be placed in.
from sklearn.model_selection import LeaveOneOut, cross_val_predict
from sklearn.ensemble import ExtraTreesRegressor, RandomForestRegressor, GradientBoostingRegressor

def _mk(nu):
    return ConstantKernel(1.0, (1e-1, 1e2)) * Matern(length_scale=[0.25, 0.25], length_scale_bounds=(1e-2, 5.0), nu=nu) \
           + WhiteKernel(noise_level=1e-2, noise_level_bounds=(1e-6, 1e0))

candidate_models = {
    'GP RBF (noisy)':   lambda: GaussianProcessRegressor(kernel_w13_check, normalize_y=True, n_restarts_optimizer=5, random_state=0),
    'GP Matern 1.5':    lambda: GaussianProcessRegressor(_mk(1.5), normalize_y=True, n_restarts_optimizer=5, random_state=0),
    'GP Matern 2.5':    lambda: GaussianProcessRegressor(_mk(2.5), normalize_y=True, n_restarts_optimizer=5, random_state=0),
    'Extra Trees':      lambda: ExtraTreesRegressor(300, random_state=0),
    'Random Forest':    lambda: RandomForestRegressor(300, random_state=0),
    'Grad. Boosting':   lambda: GradientBoostingRegressor(n_estimators=100, max_depth=2, random_state=0),
}
band = (X_updated[:, 0] > 0.65) & (X_updated[:, 0] < 0.76)
cv_results = {}
print(f"{'model':18s} {'LOO MSE (all)':>14s} {'LOO MSE (local band)':>22s}")
for name, make in candidate_models.items():
    pred = cross_val_predict(make(), X_updated, Y_updated, cv=LeaveOneOut())
    cv_results[name] = (np.mean((pred - Y_updated) ** 2), np.mean((pred[band] - Y_updated[band]) ** 2))
    print(f"{name:18s} {cv_results[name][0]:14.5f} {cv_results[name][1]:22.5f}")
winner = min(cv_results, key=lambda k: cv_results[k][0])
print(f"\nLowest overall CV error: {winner}")
print(f"Local-band RMSE for every model is about {np.sqrt(np.median([v[1] for v in cv_results.values()])):.3f}, "
      f"comparable to the whole spread of outputs in that band.")
```

**Observation:** Extra Trees has the lowest overall error, with the GP-Matern models close behind, and the original noisy RBF GP is the **worst** of the six. But inside the x1 band, where the final query has to go, every model has essentially the same error (≈0.011, RMSE ≈ 0.11). With only 22 points, the ranking is weak evidence, and no off-the-shelf surrogate can resolve a spike this narrow. Tree ensembles cannot extrapolate to a value above any observed output, so they are useless for finding a *new* maximum anyway.
The decision for the last query therefore rests on a **purpose-built local model**, not on whichever global model wins.

```python
# Three independent local estimates of where the spike's peak sits in x2 ---
# Evidence along x2 at x1 ~ 0.716 (x1 differs by <0.004 between these three points):
x2_obs = np.array([X_w10_new_point[1], X_w9_new_point[1], X_w11_new_point[1]])
y_obs  = np.array([Y_w10_new_point[0], Y_w9_new_point[0], Y_w11_new_point[0]])
print("x2 :", x2_obs);  print("y  :", np.round(y_obs, 4))

# (a) quadratic through the three points, (b) quadratic through log(y) (a Gaussian bump)
qa = np.polyfit(x2_obs, y_obs, 2);          vertex_quad  = -qa[1] / (2 * qa[0])
qb = np.polyfit(x2_obs, np.log(y_obs), 2);  vertex_gauss = -qb[1] / (2 * qb[0])

# (c) near-noise-free Matern GP fitted ONLY on the six local points, then expected improvement on a fine grid
local = (X_updated[:, 0] > 0.66) & (X_updated[:, 0] < 0.76) & (X_updated[:, 1] > 0.25) & (X_updated[:, 1] < 0.85)
X_loc, Y_loc = X_updated[local], Y_updated[local]
k_loc = ConstantKernel(0.05, (1e-3, 10)) * Matern(length_scale=[0.05, 0.05], length_scale_bounds=(5e-3, 1.0), nu=2.5) \
        + WhiteKernel(noise_level=1e-6, noise_level_bounds=(1e-8, 1e-3))
gp_local = GaussianProcessRegressor(k_loc, normalize_y=True, n_restarts_optimizer=20, random_state=1).fit(X_loc, Y_loc)
g1, g2 = np.linspace(0.69, 0.74, 101), np.linspace(0.45, 0.66, 106)
G = np.array([[a, b] for a in g1 for b in g2])
mu_l, sd_l = gp_local.predict(G, return_std=True)
best_y = Y_updated.max()
zz = (mu_l - best_y) / np.maximum(sd_l, 1e-9)
ei_l = (mu_l - best_y) * norm.cdf(zz) + sd_l * norm.pdf(zz)
pi_l = norm.cdf(zz)
peak_gp = G[np.argmax(ei_l)]

print(f"\nLocal GP uses {local.sum()} points; kernel: {gp_local.kernel_}")
print(f"(a) quadratic vertex in x2:        {vertex_quad:.4f}   (peak value extrapolates to {np.polyval(qa, vertex_quad):.3f} -- trust the location, not the height)")
print(f"(b) log-quadratic vertex in x2:    {vertex_gauss:.4f}")
print(f"(c) local-GP max expected improvement at x = {peak_gp}  (EI={ei_l.max():.4f}, P(improve)={pi_l[np.argmax(ei_l)]:.2f})")
```

**Option A (refine the spike):** The three local estimates above put the peak at x2 ≈ 0.58–0.60 at Week 9's x1. This option follows the one place
where the data have shown a value above 0.7. Its weakness is that it rests on three nearly collinear points and a local GP whose length-scale is pinned
at its 0.005 lower bound, so its probabilities are fragile.

**Option B (fill the largest gap):** Inside the proven x1 band, the longest stretch of x2 with no observation is x2 ≈ 0.30–0.55 (width 0.25). A
space-filling rule (query the midpoint of the largest empty interval) picks a point with no tuning at all. Before choosing between A and B, the next
cell asks what **my own 22 observations** say about each.

**Data check before comparing the options:** Along x2 inside the proven x1 band, points that are *close* in x2 (lag < 0.06) differ by as much as, or more than, points that are far apart: the semivariance at the shortest lag (≈0.011) is higher than at every longer lag (≈0.005–0.009) and is about 1.5× the band's overall variance (0.0073). With only 12 band points and six short-lag pairs, all from the Week 9–11 cluster, this is weak evidence, but it agrees with the saturated x2 length-scale found in Week 12: **an observation next to a candidate does not make the candidate's outcome more predictable.** So a query close to Week 9 is not 'safer' than one in an unexplored stretch; each is close to an independent draw from the band's output distribution (mean ≈ 0.56, std ≈ 0.09).

```python
# Data check of how far a nearby observation tells me anything, along x2, inside the proven x1 band 
# Empirical semivariance: half the squared output difference between two band points, grouped by their x2 distance.
# If neighbours really were informative, short-lag semivariance would be clearly BELOW long-lag semivariance.
inb_ = (X_updated[:, 0] > 0.65) & (X_updated[:, 0] < 0.76)
xb_, yb_ = X_updated[inb_], Y_updated[inb_]
ii, jj = np.triu_indices(len(yb_), 1)
lag = np.abs(xb_[ii, 1] - xb_[jj, 1]); semiv = 0.5 * (yb_[ii] - yb_[jj]) ** 2
print(f"{len(yb_)} band points; output mean {yb_.mean():.3f}, std {yb_.std(ddof=1):.3f}; overall variance {yb_.var(ddof=1):.4f}")
print(f"{'x2 lag':>12s} {'pairs':>6s} {'semivariance':>13s}")
for lo, hi in [(0, 0.06), (0.06, 0.15), (0.15, 0.30), (0.30, 0.60), (0.60, 1.01)]:
    sel = (lag >= lo) & (lag < hi)
    print(f"{lo:5.2f}-{hi:4.2f}   {sel.sum():6d} {semiv[sel].mean():13.4f}")
print(f"\nOutputs in the band that reach the current best: {(yb_ >= Y_updated.max() - 1e-12).sum()} of {len(yb_)}  (that single one is Week 9's point)")
```

```python
# Option B coordinates from a space-filling rule on MY OWN data (largest empty x2-interval in the proven x1 band) ---
band_pts = X_updated[(X_updated[:, 0] > 0.65) & (X_updated[:, 0] < 0.76)]
x2_sorted = np.sort(np.unique(np.round(band_pts[:, 1], 6)))
gaps = np.diff(x2_sorted)
k = int(np.argmax(gaps))
x2_gap_mid = float(round((x2_sorted[k] + x2_sorted[k + 1]) / 2, 3))
x1_band_centre = float(round(np.median(band_pts[Y_updated[(X_updated[:, 0] > 0.65) & (X_updated[:, 0] < 0.76)] > 0.5, 0]), 3))
print(f"x2 values observed in the x1 band: {np.round(x2_sorted, 3)}")
print(f"Largest empty x2-interval: ({x2_sorted[k]:.3f}, {x2_sorted[k+1]:.3f}), width {gaps[k]:.3f}, midpoint {x2_gap_mid}")
print(f"x1 = median x1 of the band points that scored > 0.5: {x1_band_centre}")

point_A = np.array([X_w9_new_point[0], round(float(np.mean([vertex_quad, vertex_gauss, peak_gp[1]])), 3)])   # refine the spike
point_B = np.array([x1_band_centre, x2_gap_mid])                                                               # fill the largest gap

# --- what do my own surrogates say about each option? probability of beating the current best (PI) ---
gp_m15 = GaussianProcessRegressor(_mk(1.5), normalize_y=True, n_restarts_optimizer=10, random_state=42).fit(X_updated, Y_updated)
gp_m25 = GaussianProcessRegressor(_mk(2.5), normalize_y=True, n_restarts_optimizer=10, random_state=42).fit(X_updated, Y_updated)
surrogates = {'global GP, RBF (noisy)': gpr_w13, 'global GP, Matern 1.5': gp_m15, 'global GP, Matern 2.5': gp_m25, 'local GP (6 pts, near noise-free)': gp_local}

def pi_ei(gp, pt):
    m, s = gp.predict(pt.reshape(1, -1), return_std=True)
    z = (m[0] - best_y) / max(s[0], 1e-9)
    return m[0], s[0], norm.cdf(z), (m[0] - best_y) * norm.cdf(z) + s[0] * norm.pdf(z)

print(f"\nOption A = {point_A}   Option B = {point_B}   (current best y = {best_y:.4f})")
print(f"{'surrogate':36s} {'A: mu':>7s} {'A: P(beat)':>11s} {'B: mu':>7s} {'B: P(beat)':>11s}")
tab = {}
for nm, gp in surrogates.items():
    a, b = pi_ei(gp, point_A), pi_ei(gp, point_B); tab[nm] = (a, b)
    print(f"{nm:36s} {a[0]:7.3f} {a[2]:11.3f} {b[0]:7.3f} {b[2]:11.3f}")

# how much does ANY point in the band stand out under the global models?
Gb = np.array([[a, b] for a in np.linspace(0.66, 0.75, 19) for b in np.linspace(0.0, 1.0, 41)])
pi_all = np.mean([norm.cdf((gp.predict(Gb)-best_y)/gp.predict(Gb, return_std=True)[1]) for gp in list(surrogates.values())[:3]], axis=0)
print(f"\nGlobal models: P(beat best) ranges only {pi_all.min():.3f}-{pi_all.max():.3f} across the whole x1 band -- they cannot tell the candidates apart.")
```

**Decision:** The three global surrogates give Option A and Option B almost the same chance of beating 0.711209 (**≈ 6% each**, across the entire band only about 1–8%, the highest being on the x2 = 1 edge, which is extrapolation). The local near-noise-free GP separates them only slightly (about 9.0% for A vs 8.4% for B) and its length-scale is pinned at a bound. 
The semivariance check shows closeness to a known point buys no predictability. Evidence is therefore **indifferent** between the two, and I break the tie with a rule that needs no model:
**query the point farthest from everything already observed inside the proven band** (the midpoint of its largest empty x2-interval). Reasons that follow from data alone: <br>
1. Option A is within 0.015 of Week 9, and the three queries there returned 0.711, 0.639 and 0.453, so that neighbourhood has already been explored thrice<br>
2. The gap x2 ≈ 0.30–0.55 holds a single point at its edge, far too little to rule it out, as the Week 9–11 cluster showed how quickly outputs change<br>
3. Grading uses the best pair found, so a draw from a region I have barely sampled cannot lower the record best.

# Week 13 Outcome and Final Refit
The final query (0.701, 0.426) has now been run and returned **0.649151**. This section compares that result with the forecast made *before* seeing it, checks whether the record best changed, and refits the surrogates on all **23** points.

```python
# Week 13 outcome: forecast vs result, new-best check, refit on all 23 points
z13 = (Y_w13_new_point[0] - mu_13[0]) / sd_13[0]
print(f"Pre-result forecast at {X_w13_new_point}: mu={mu_13[0]:.4f}, std={sd_13[0]:.4f}")
print(f"Actual outcome: {Y_w13_new_point[0]:.4f}  ->  {z13:+.2f} predicted std devs from the forecast mean")
print(f"Change vs the recorded best ({best_y:.6f}): {Y_w13_new_point[0] - best_y:+.4f}  -> new best? {Y_w13_new_point[0] > best_y}")
i_best = int(np.argmax(Y_updated))
print(f"Best pair on all {len(Y_updated)} points: x = {X_updated[i_best]}, y = {Y_updated[i_best]:.6f}")

band23 = (X_updated[:, 0] > 0.65) & (X_updated[:, 0] < 0.76)
print(f"\nx1 band (0.65-0.76) now holds {band23.sum()} points: mean {Y_updated[band23].mean():.3f}, std {Y_updated[band23].std(ddof=1):.3f}"
      f"  (before Week 13: mean {Y22[(X22[:,0]>0.65)&(X22[:,0]<0.76)].mean():.3f})")

# refit the global GP on all 23 points (same kernel family as before)
gpr_final = GaussianProcessRegressor(kernel=kernel_w13_check, normalize_y=True, n_restarts_optimizer=10, random_state=42)
gpr_final.fit(X_updated, Y_updated)
print(f"\nKernel on all 23 points: {gpr_final.kernel_}")

# leave-one-out bake-off repeated on 23 points for the final dashboard
cv_final = {}
for name, make in candidate_models.items():
    pred = cross_val_predict(make(), X_updated, Y_updated, cv=LeaveOneOut())
    cv_final[name] = (np.mean((pred - Y_updated) ** 2), np.mean((pred[band23] - Y_updated[band23]) ** 2))
winner_final = min(cv_final, key=lambda k: cv_final[k][0])
print(f"\n{'model':18s} {'LOO MSE (all 23)':>17s} {'LOO MSE (band)':>15s}")
for n, v in cv_final.items(): print(f"{n:18s} {v[0]:17.5f} {v[1]:15.5f}")
print(f"Lowest overall CV error on 23 points: {winner_final}")
```

**Observation:** The final query returned a value inside the forecast's one-sigma band and above its mean, but it did not beat the record, so the best pair is still Week 9's. Together with Week 12 (0.534) this is a second reading from the x2 gap of the proven band: both are ordinary values (≈0.53–0.65), not a second spike. Two points still cannot rule out a narrow peak between them, and with no queries left that question stays open. The cost of the final query was zero for the recorded result, as the grading rule (best pair found) implied.
