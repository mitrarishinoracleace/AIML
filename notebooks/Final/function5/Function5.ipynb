# Bug Fix ( Week 4 )
Week 3's output was **-1.359780**, which is **below -1**. np.log1p(y) = log(1 + y) is only defined for y > -1; for y <= -1, 1 + y <= 0 and the log is undefined.
Hence, np.log1p(-1.36) returns **NaN**. Since log1p(Y) was modelled since Week 1, this would break the surrogate fit from this point on.

**Fix**: switch to a *signed*-log transform, which compresses large magnitudes the same way log1p did but is defined for the full real line:
```
signed_log1p(y) = sign(y) * log1p( | y | )
inverse         = sign(z) * expm1( | z | )
```
This fix is applied everywhere log1p/expm1 were previously used: the main GP fit below, predict_yield() and inside bayesian_optimization_step().

````python
def signed_log1p(y):
    """Like log1p, but defined for the full real line (handles y <= -1)."""
    return np.sign(y) * np.log1p(np.abs(y))

def inverse_signed_log1p(z):
    """Inverse of signed_log1p."""
    return np.sign(z) * np.expm1(np.abs(z))

# Demonstrate the bug and the fix on Week 3's actual output:
print("Old np.log1p(-1.359780)      :", np.log1p(-1.359780271551876), " <- NaN, this is the bug")
print("New signed_log1p(-1.359780)  :", signed_log1p(-1.359780271551876))
print("Round-trip check             :", inverse_signed_log1p(signed_log1p(-1.359780271551876)))
````

# Compute the UCB acquisition function

- UCB(x)=μ(x)+β⋅σ(x)
- A high beta favors exploration (uncertainty) over exploitation (predicted mean).
- Week 1 uses beta = 2.5, reflecting the 90:10 exploration-heavy strategy.

````python
# Function to predict mean and std for raw x
def predict_yield(X_raw):
    Xs          = X_scaler.transform(X_raw)
    mu_s, std_s = gpr.predict(Xs, return_std = True)
    mu_log      = mu_s * Y_std + Y_mean
    std_log     = std_s * Y_std
    #mu_yield    = np.expm1(mu_log)    # Bug Fix ( Week 4 )
    mu_yield    = inverse_signed_log1p(mu_log)
    return mu_yield, std_log

# Setting a higher beta(= 2.5) for Week -1
#beta         = 2.5                

# Setting a higher beta(= 1.8) for Week -8
beta         = 1.8  

# Function to calculare UCB 
def ucb_log_space(X_raw, beta = beta):
    Xs          = X_scaler.transform(X_raw)
    mu_s, std_s = gpr.predict(Xs, return_std = True)
    return mu_s + beta * std_s

# Commented Code for Week 1 
# n_candidates = 200000
# RNG          = np.random.default_rng(42)
# candidates   = RNG.uniform(0, 1, size = (n_candidates, 4))

# ucb_scores   = ucb_log_space(candidates)
# next_idx     = np.argmax(ucb_scores)
# x_next       = candidates[next_idx]

# pred_mu, pred_std_log = predict_yield(x_next.reshape(1, -1))

# print(f"Suggested next Query Point: {x_next}")
# print(f"Predicted Mean(μ): {pred_mu[0]:.6e}")
# print(f"Predicted Std (σ): {pred_std_log[0]:.6e}")
# print(f"UCB Score: {ucb_scores[next_idx]:.6e}")
````

# Week 2: Consolidated Bayesian Optimisation Function
Define a consolidated function that includes acquisition optimisation into a single reusable function. Two changes from Week 1's approach:

- **Multi-start L-BFGS-B** on the negative UCB, instead of scoring 200,000 random candidates and taking the arg-max. This locates the acquisition optimum more precisely, as the search area becomes smaller.
- The function accepts explicit **bounds**, so it can be reused for the narrowed region this week and for the full `[0,1)^4` cube in earlier/future weeks alike.

````python
def suggest_next_query(gpr, X_scaler, Y_mean, Y_std, bounds, beta = 1.8, n_restarts = 25, seed = 42):
    rng      = np.random.default_rng(seed)
    dim      = bounds.shape[0]

    def neg_acq(x):
        x           = x.reshape(1, -1)
        Xs          = X_scaler.transform(x)
        mu_s, std_s = gpr.predict(Xs, return_std = True)
        return -(mu_s[0] + beta * std_s[0])

    best_val, best_x = np.inf, None
    starts           = rng.uniform(bounds[:, 0], bounds[:, 1], size = (n_restarts, dim))
    for x0 in starts:
        res          = minimize(neg_acq, x0, bounds = list(zip(bounds[:, 0], bounds[:, 1])), method = "L-BFGS-B")
        if res.fun < best_val:
            best_val, best_x = res.fun, res.x

    mu_next, std_next = predict_yield(best_x.reshape(1, -1))
    return best_x, mu_next[0], std_next[0], -best_val
````

# Week 3: Consolidated Bayesian Optimisation Function

````python
def bayesian_optimization_step(X, Y, bounds, beta = 1.6, kernel_type = "rbf", n_restarts_gp = 10, n_restarts_acq = 25, seed = 42):
    X_scaler_l = StandardScaler()
    X_s_l      = X_scaler_l.fit_transform(X)
    #Y_log_l    = np.log1p(Y)     # Commented for Bug Fix ( Week 4 )
    Y_log_l    = signed_log1p(Y)  # Added for Week 4 Bug Fix
    Y_mean_l, Y_std_l = Y_log_l.mean(), Y_log_l.std()
    Y_s_l      = (Y_log_l - Y_mean_l) / Y_std_l

    if kernel_type == "rbf":
        kernel_l = ConstantKernel(1.0, (1e-2, 1e2)) * RBF(length_scale = [1.0] * X.shape[1], length_scale_bounds = (1e-2, 1e2)) \
                   + WhiteKernel(noise_level = 1e-2, noise_level_bounds = (1e-5, 1e0))
    elif kernel_type == "matern":
        kernel_l = ConstantKernel(1.0, (1e-2, 1e2)) * Matern(length_scale = [1.0] * X.shape[1], length_scale_bounds = (1e-2, 1e2), nu = 1.5) \
                   + WhiteKernel(noise_level = 1e-2, noise_level_bounds = (1e-5, 1e0))
    else:
        raise ValueError("kernel_type must be 'rbf' or 'matern'")

    gpr_l = GaussianProcessRegressor(kernel = kernel_l, normalize_y = False, n_restarts_optimizer = n_restarts_gp, random_state = seed)
    gpr_l.fit(X_s_l, Y_s_l)
    lml_l = gpr_l.log_marginal_likelihood(gpr_l.kernel_.theta)

    def predict_yield_l(X_raw):
        Xs = X_scaler_l.transform(X_raw)
        mu_s, std_s = gpr_l.predict(Xs, return_std = True)
        mu_log      = mu_s * Y_std_l + Y_mean_l
        std_log     = std_s * Y_std_l
        #return np.expm1(mu_log), std_log              # Commented for Bug Fix ( Week 4 )
        return inverse_signed_log1p(mu_log), std_log   # Added for Bug Fix ( Week 4 )

    rng = np.random.default_rng(seed)
    dim = bounds.shape[0]

    def neg_acq(x):
        x = x.reshape(1, -1)
        Xs = X_scaler_l.transform(x)
        mu_s, std_s = gpr_l.predict(Xs, return_std = True)
        return -(mu_s[0] + beta * std_s[0])

    best_val, best_x = np.inf, None
    starts = rng.uniform(bounds[:, 0], bounds[:, 1], size = (n_restarts_acq, dim))
    for x0 in starts:
        res = minimize(neg_acq, x0, bounds = list(zip(bounds[:, 0], bounds[:, 1])), method = "L-BFGS-B")
        if res.fun < best_val:
            best_val, best_x = res.fun, res.x

    mu_next, std_next = predict_yield_l(best_x.reshape(1, -1))

    return {
        "kernel_type": kernel_type, "gpr": gpr_l, "X_scaler": X_scaler_l,
        "Y_mean": Y_mean_l, "Y_std": Y_std_l, "log_marginal_likelihood": lml_l,
        "x_next": best_x, "mu_next": mu_next[0], "std_next": std_next[0],
        "ucb_next": -best_val,
    }

# --- Narrow the search region around the CURRENT best (now Week 2's point) ---
delta      = 0.2                                     # Week 3 delta reduced from 0.25 to 0.2
best_idx   = np.argmax(Y)                            # recompute in case data changed above
best_x_now = X[best_idx]

lb         = np.clip(best_x_now - delta, 0.0, 0.999999)
ub         = np.clip(best_x_now + delta, 0.0, 0.999999)
bounds     = np.vstack([lb, ub]).T
print("Week 3 narrowed search bounds (per dimension):\n", bounds)

beta = 1.6                                           # Week 3: beta reduced from 1.8 to 1.6 to facilitate exploitation

# Check Kernel best-fit
result_rbf    = bayesian_optimization_step(X, Y, bounds, beta = beta, kernel_type = "rbf")
result_matern = bayesian_optimization_step(X, Y, bounds, beta = beta, kernel_type = "matern")

print(f"bayesian_optimization_step RBF LML     : {result_rbf['log_marginal_likelihood']:.3f}")
print(f"bayesian_optimization_step Matern LML  : {result_matern['log_marginal_likelihood']:.3f}")

final = result_matern if result_matern["log_marginal_likelihood"] > result_rbf["log_marginal_likelihood"] else result_rbf


gpr             = final["gpr"]
X_scaler        = final["X_scaler"]
Y_mean          = final["Y_mean"]
Y_std           = final["Y_std"]
x_next          = final["x_next"]
pred_mu         = final["mu_next"]
pred_std_log    = final["std_next"]
ucb_next        = final["ucb_next"]

print(f"\nWeek 3 chosen kernel      : {final['kernel_type']}")
print(f"Suggested next Query Point  : {np.round(x_next, 6)}")
print(f"Predicted Mean (mu)         : {pred_mu:.4f}")
print(f"Predicted Std (sigma, log)  : {pred_std_log:.4f}")
print(f"UCB Score                   : {ucb_next:.4f}")
````

# Week 4: Implementing LOO cross-validation and a bracketed search around the discovered cliff

````python
# Leave-one-out cross-validation of the GP 
def loo_cv_gp(X, Y, kernel_type = "rbf", seed = 42):
    n                = len(Y)
    residuals, in_ci = [], []
    for i in range(n):
        mask             = np.ones(n, dtype = bool)
        mask[i]          = False
        X_train, Y_train = X[mask], Y[mask]
        X_test, Y_test   = X[i:i+1], Y[i]

        Xsc              = StandardScaler()
        Xs_tr            = Xsc.fit_transform(X_train)
        Ylog_tr          = signed_log1p(Y_train)
        ymean_l, ystd_l  = Ylog_tr.mean(), Ylog_tr.std()
        Ys_tr            = (Ylog_tr - ymean_l) / ystd_l

        if kernel_type == "rbf":
            k            = ConstantKernel(1.0, (1e-2, 1e2)) * RBF(length_scale = [1.0] * X.shape[1], length_scale_bounds = (1e-2, 1e2)) \
                             + WhiteKernel(noise_level = 1e-2, noise_level_bounds = (1e-5, 1e0))
        else:
            k            = ConstantKernel(1.0, (1e-2, 1e2)) * Matern(length_scale = [1.0] * X.shape[1], length_scale_bounds = (1e-2, 1e2), nu = 1.5) \
                + WhiteKernel(noise_level = 1e-2, noise_level_bounds = (1e-5, 1e0))

        g                = GaussianProcessRegressor(kernel = k, normalize_y = False, n_restarts_optimizer = 3, random_state = seed)
        g.fit(Xs_tr, Ys_tr)

        mu_s, std_s      = g.predict(Xsc.transform(X_test), return_std = True)
        mu_log           = mu_s[0] * ystd_l + ymean_l
        std_log          = std_s[0] * ystd_l
        mu_pred          = inverse_signed_log1p(mu_log)

        residuals.append(Y_test - mu_pred)
        half_width       = abs(inverse_signed_log1p(mu_log + std_log) - mu_pred)
        in_ci.append(mu_pred - 1.96*half_width <= Y_test <= mu_pred + 1.96 * half_width)

    residuals            = np.array(residuals)
    rmse                 = np.sqrt(np.mean(residuals**2))
    coverage             = np.mean(in_ci)
    return rmse, coverage, residuals

rmse_gp, coverage_gp, resid_gp = loo_cv_gp(X, Y, kernel_type = "rbf")
print(f"GP  LOO-CV RMSE               : {rmse_gp:.3f}")
print(f"GP  LOO-CV 95% CI coverage   : {coverage_gp*100:.1f}%  (target 95%)")

# Checking if a Neural Network applicable could be a fit
def loo_cv_mlp(X, Y, seed = 42):
    n           = len(Y)
    residuals   = []
    for i in range(n):
        mask    = np.ones(n, dtype = bool)
        mask[i] = False
        X_train, Y_train = X[mask], Y[mask]
        X_test, Y_test   = X[i:i+1], Y[i]

        Xsc    = StandardScaler()
        Xs_tr = Xsc.fit_transform(X_train)
        Ylog_tr = signed_log1p(Y_train)
        ymean_l, ystd_l = Ylog_tr.mean(), Ylog_tr.std()
        Ys_tr = (Ylog_tr - ymean_l) / ystd_l

        mlp = MLPRegressor(hidden_layer_sizes = (8,), max_iter = 5000, random_state = seed)
        mlp.fit(Xs_tr, Ys_tr)

        pred_s  = mlp.predict(Xsc.transform(X_test))[0]
        mu_log  = pred_s * ystd_l + ymean_l
        mu_pred = inverse_signed_log1p(mu_log)
        residuals.append(Y_test - mu_pred)
    residuals = np.array(residuals)
    return np.sqrt(np.mean(residuals**2)), residuals

rmse_mlp, resid_mlp = loo_cv_mlp(X, Y)
print(f"\nMLP (small NN) LOO-CV RMSE    : {rmse_mlp:.3f}")
print(f"GP             LOO-CV RMSE    : {rmse_gp:.3f}")
winner = "GP" if rmse_gp < rmse_mlp else "MLP"
print(f"-> {winner} generalises better at n={len(Y)} points -- confirms the GP is still the right surrogate for now.")

# Bracket the discovered cliff along x2, hence keeping x1/x3/x4 tight near the boundary
good_pt = X_w2_new_point   # x2=0.596 -> 1909.16
bad_pt  = X_w3_new_point   # x2=0.796 -> -1.36

lb = np.array([0.0 , min(good_pt[1], bad_pt[1]), 0.95, 0.95])
ub = np.array([0.05, max(good_pt[1], bad_pt[1]), 0.999999, 0.999999])
bounds = np.vstack([lb, ub]).T
print("\nWeek 4 bracketed search bounds (per dimension):\n", bounds)

beta = 1.0   # WEEK 4 CHANGE: 1.6 -> 1.0 (favour precise localisation of the cliff over broad exploration)

result_rbf_w4    = bayesian_optimization_step(X, Y, bounds, beta = beta, kernel_type = "rbf")
result_matern_w4 = bayesian_optimization_step(X, Y, bounds, beta = beta, kernel_type = "matern")
print(f"\nWeek 4 RBF    LML: {result_rbf_w4['log_marginal_likelihood']:.3f}")
print(f"Week 4 Matern LML: {result_matern_w4['log_marginal_likelihood']:.3f}")

final_w4 = result_matern_w4 if result_matern_w4["log_marginal_likelihood"] > result_rbf_w4["log_marginal_likelihood"] else result_rbf_w4

gpr             = final_w4["gpr"]
X_scaler        = final_w4["X_scaler"]
Y_mean          = final_w4["Y_mean"]
Y_std           = final_w4["Y_std"]
x_next          = final_w4["x_next"]
pred_mu         = final_w4["mu_next"]
pred_std_log    = final_w4["std_next"]
ucb_next        = final_w4["ucb_next"]

print(f"\nWeek 4 chosen kernel        : {final_w4['kernel_type']}")
print(f"Suggested next Query Point  : {np.round(x_next, 6)}")
print(f"Predicted Mean (mu)         : {pred_mu:.4f}")
print(f"Predicted Std (sigma, log)  : {pred_std_log:.4f}")
print(f"UCB Score                   : {ucb_next:.4f}")
````

# Week 12 Change: corner has plateaued - use the penultimate query to probe, keep the final one for exploitation

**Observation from Week 11's result:** the tight confirmatory query returned **1915.346** (GP predicted ~1914.6), 0.6 below the Week 10 best. - The best refinement candidate has a predicted mean (1914.6) *below* the current best and an expected improvement of only 0.003 log-units (0.3%).
- Week 13 is the final submission, so Week 12 is the last query where an unsuccessful result costs nothing (the 1915.96 point is already banked, assuming the best output found is what counts - if only the last query is judged, set `WEEK12_MODE = "exploit"` below).
- Only one region away from the corner has ever been tested directly (Week 6, a poor 68.5, taken ~0.6 from the best original data point). The three best original points (best: 0.224, 0.846, 0.879, 0.879 -> 1088.9) all have x2 in 0.77-0.86 and x4 >= 0.85, which suggests a second, untested ridge.

**Changes:**

2. **Beta**: 0.35 -> **1.0**. The sweep below shows that over the full domain the UCB maximiser lands in the second region at *every* beta from 0.35 to 2.0 (drifting towards x4 = 1 as beta grows) - it was the narrow box, not beta, that kept earlier weeks at the corner. At beta 0.35 the predicted mean there (2270) is an extrapolation artefact twice the best original value; 1.0 is a mid-range value with a more modest prediction (1870) that is stable across the sweep.
3. Both candidates (exploit and probe) are computed and compared on predicted mean, uncertainty, probability of improvement (PI) and expected improvement (EI). **Caveat:** the GP's variance far from data is large by construction, so PI/EI for the probe are optimistic; realistically, one probe in 4D is more likely to miss (as Week 6 did) than to win.
4. **Decision rule for Week 13 (final):** if the Week 12 result is above 1500 refine around it (low beta, tight box); otherwise re-submit the corner (x2 ~ 0.5997).

````python
# Sanity check: Checking how the Week 11 GP predicted Week 11's query
mu_check, std_check = predict_yield(X_w11_new_point.reshape(1, -1))
print(f"GP prediction at Week 11's query : mu = {mu_check[0]:.3f}  (true y = {Y_w11_new_point[0]:.3f})")
print(f"Error (true - predicted)         : {Y_w11_new_point[0] - mu_check[0]:.3f}")

# Plateau diagnostic - checkig how much do the corner queries still differ from one another
corner_mask = (X[:, 0] < 0.001) & (X[:, 2] > 0.99) & (X[:, 3] > 0.99) & (Y > 1000)
plateau     = np.sort(Y[corner_mask])[::-1]
print(f"\nCorner queries (x1~0, x3/x4~1, y>1000): n = {corner_mask.sum()}")
print("  y values (desc):", np.round(plateau, 2))
print(f"  spread (max - min) = {plateau.max() - plateau.min():.2f}  ({(plateau.max() - plateau.min()) / plateau.max() * 100:.2f}% of best)")

# Re-running RBF vs Matern comparison on the updated (31-point) dataset
kernel_matern_w12 = ConstantKernel(1.0, (1e-2, 1e2)) * Matern(length_scale = [1.0] * 4, length_scale_bounds = (1e-2, 1e2), nu = 1.5) \
                    + WhiteKernel(noise_level = 1e-2, noise_level_bounds = (1e-5, 1e0))
gpr_matern_w12 = GaussianProcessRegressor(kernel = kernel_matern_w12, normalize_y = False, n_restarts_optimizer = 10, random_state = 42)
gpr_matern_w12.fit(X_s, Y_s)

lml_rbf_w12    = gpr.log_marginal_likelihood(gpr.kernel_.theta)
lml_matern_w12 = gpr_matern_w12.log_marginal_likelihood(gpr_matern_w12.kernel_.theta)
print(f"\nWeek 12 RBF    LML: {lml_rbf_w12:.3f}")
print(f"Week 12 Matern LML: {lml_matern_w12:.3f}")
kernel_w12 = "rbf" if lml_rbf_w12 >= lml_matern_w12 else "matern"
print(f"Kernel choice: {kernel_w12.upper()}")

# Full-dataset LOO-CV re-check
rmse_gp_w12, coverage_gp_w12, _ = loo_cv_gp(X, Y, kernel_type = kernel_w12)
print(f"\nWeek 12 GP LOO-CV RMSE             : {rmse_gp_w12:.3f}  (Week 11 was {rmse_gp_w11:.3f})")
print(f"Week 12 GP LOO-CV ~95% CI coverage : {coverage_gp_w12*100:.1f}%  (Week 11 was {coverage_gp_w11*100:.1f}%, target 95%)")

# WEEK 12 CHANGE: beta sweep over the FULL domain - where does the UCB maximiser land as beta grows?
full_bounds = np.array([[0.0, 0.999999]] * 4)
sweep_betas = [0.35, 0.5, 0.75, 1.0, 1.5, 2.0]
sweep_rows  = []
print("\nBeta sweep (full domain): where does the UCB maximiser land?")
for b in sweep_betas:
    r = bayesian_optimization_step(X, Y, full_bounds, beta = b, kernel_type = kernel_w12)
    sweep_rows.append((b, r["x_next"], r["mu_next"], r["std_next"]))
    print(f"  beta = {b:4.2f}  x = {np.round(r['x_next'], 3)}  mu = {r['mu_next']:8.1f}  sigma(log) = {r['std_next']:.3f}")

# WEEK 12 CHANGE: the two candidates, side by side
beta_exploit, beta_probe = 0.35, 1.0
corner_bounds = np.array([[0.0, 0.0008], [0.57, 0.63], [0.998, 0.999999], [0.998, 0.999999]])   # Week 11's box, for the exploit candidate
result_exploit = bayesian_optimization_step(X, Y, corner_bounds, beta = beta_exploit, kernel_type = kernel_w12)
result_probe   = bayesian_optimization_step(X, Y, full_bounds,   beta = beta_probe,   kernel_type = kernel_w12)

def improvement_stats(res, y_best):
    # probability / expected improvement over the current best, in the GP's standardised log space
    mu_s, sd_s = res["gpr"].predict(res["X_scaler"].transform(res["x_next"].reshape(1, -1)), return_std = True)
    best_s     = (signed_log1p(y_best) - res["Y_mean"]) / res["Y_std"]
    z          = (mu_s[0] - best_s) / sd_s[0]
    return norm.cdf(z), sd_s[0] * (z * norm.cdf(z) + norm.pdf(z)) * res["Y_std"]    # PI, EI (log units)

y_best_now = Y.max()
cand_rows  = {}
for name, res in [("exploit (corner, beta=0.35)", result_exploit), ("probe (full domain, beta=1.0)", result_probe)]:
    pi, ei = improvement_stats(res, y_best_now)
    cand_rows[name] = (res, pi, ei)
    print(f"\n{name}")
    print(f"  x = {np.round(res['x_next'], 6)}   mu = {res['mu_next']:.1f}   sigma(log) = {res['std_next']:.3f}   PI = {pi:.3f}   EI = {ei:.4f} (log units)")

# WEEK 12 CHANGE: choose which candidate to submit (switch back to "exploit" to re-submit the corner refinement instead)
WEEK12_MODE = "probe"
final_w12   = result_probe if WEEK12_MODE == "probe" else result_exploit
beta        = beta_probe   if WEEK12_MODE == "probe" else beta_exploit
bounds      = full_bounds  if WEEK12_MODE == "probe" else corner_bounds
best_idx    = np.argmax(Y)
best_x_now  = X[best_idx]                                # still Week 10's point

gpr             = final_w12["gpr"]
X_scaler        = final_w12["X_scaler"]
Y_mean          = final_w12["Y_mean"]
Y_std           = final_w12["Y_std"]
x_next          = final_w12["x_next"]
pred_mu         = final_w12["mu_next"]
pred_std_log    = final_w12["std_next"]
ucb_next        = final_w12["ucb_next"]

print(f"\nWeek 12 mode                : {WEEK12_MODE}")
print(f"Week 12 chosen kernel       : {final_w12['kernel_type']}")
print(f"Suggested next Query Point  : {np.round(x_next, 6)}")
print(f"Predicted Mean (mu)         : {pred_mu:.4f}")
print(f"Predicted Std (sigma, log)  : {pred_std_log:.4f}")
print(f"UCB Score                   : {ucb_next:.4f}")
````

# Week 13 (Final): Test the unexplored high-x basin for a second maximum

**Observation from Week 12's result:** the full-domain probe returned **1085.354**, almost exactly the level of the best original data point (0.224, 0.846, 0.879, 0.879 -> 1088.9). A second region at ~1085-1089 therefore exists, but it is lower than the corner (1915.96). It did not clear the ~1500 threshold set last week for refining it.

**Correction to the Week 11-12 reasoning:** those weeks called the corner queries a noisy plateau. They are not: yield is linear in x2 over the five corner queries (y = 2010 * x2 + 710, R^2 = 0.999995), so the function is deterministic and the corner is a *ramp* that was still rising when sampled. The Week 3 'cliff' (y = -1.36 at x2 = 0.796) is a single point: it shows a dip at x2 = 0.8, not what happens beyond it.

**Hypothesis for the final query:** a second, higher maximum may sit in the basin that no query has reached - the far side of the x2 dip, where x2, x3 and x4 are all near their maximum. Evidence from my own data:
1. The corner is rising in x2 (slope ~2010, slightly increasing) and x3, x4 are already at their maximum there.
2. Across all 32 points, x3 and x4 correlate positively with yield (computeirection.
3. No query lies near the far vertex (nearest observed point is ~0.78 away), and the GP gives most unexplored vertices the same maximum uncertainty, so the model cannot rank them - this needs a deliberate probe, not an acquisition-function maximiser (which proposes interior y dataset.

**Decision:**
- Final query = the far vertex **(0.999999, 0.999999, 0.999999, 0.999999)**, a hypothesis test for a second maximum.
- The bounded step along the measured x2 ramp (x2 = 0.602, expected ~1920) stays available via `FINAL_MODE = "ramp"`.
- If the result is above 1915.96 it becomes the best found; if not, the best remains the Week 10 point. If only the final query is judged rather than the best result found, `FINAL_MODE = "ramp"` is the lower-risk choice.

````python
from scipy.stats import norm
import itertools

X_pre, Y_pre = X[:-1], Y[:-1]   

# 1. Out-of-sample check of the Week 12 probe: refit the GP on the 31 points available BEFORE the probe
n_before = len(Y_pre) - 1                                   # Week 12's point is the last row of X / Y
Xsc_b    = StandardScaler();  Xs_b = Xsc_b.fit_transform(X[:n_before])
Ylog_b   = signed_log1p(Y[:n_before]);  ym_b, ys_b = Ylog_b.mean(), Ylog_b.std()
kernel_b = ConstantKernel(1.0, (1e-2, 1e2)) * RBF(length_scale = [1.0] * 4, length_scale_bounds = (1e-2, 1e2)) \
           + WhiteKernel(noise_level = 1e-2, noise_level_bounds = (1e-5, 1e0))
gpr_b    = GaussianProcessRegressor(kernel = kernel_b, normalize_y = False, n_restarts_optimizer = 10, random_state = 42)
gpr_b.fit(Xs_b, (Ylog_b - ym_b) / ys_b)
mu_s, sd_s = gpr_b.predict(Xsc_b.transform(X_w12_new_point.reshape(1, -1)), return_std = True)
mu_log_b, sd_log_b = mu_s[0] * ys_b + ym_b, sd_s[0] * ys_b
print(f"Week 12 probe, out-of-sample: predicted {inverse_signed_log1p(mu_log_b):.1f}  (log-space sigma = {sd_log_b:.2f}),  actual {Y_w12_new_point[0]:.1f}")
print(f"  actual is {(signed_log1p(Y_w12_new_point[0]) - mu_log_b) / sd_log_b:+.2f} sigma from the prediction (log space) -> inside the model's own error bars")
best_orig_i = np.argmax(Y0)
print(f"  distance from the probe to the best ORIGINAL point {np.round(X0[best_orig_i], 3)} (y = {Y0[best_orig_i]:.1f}): {np.linalg.norm(X_w12_new_point - X0[best_orig_i]):.3f}")
print(f"  probe y = {Y_w12_new_point[0]:.1f} vs original-point y = {Y0[best_orig_i]:.1f}: a second region at ~1085-1089, lower than the corner ({Y_pre.max():.1f})")

# 2. The corner ramp: yield vs x2 for the corner queries (x1 ~ 0, x3 = x4 = 0.999999, y > 1000)
ramp_mask   = (X_pre[:, 0] < 0.001) & (X_pre[:, 2] > 0.99) & (X_pre[:, 3] > 0.99) & (Y_pre > 1000)
ramp_x2     = X_pre[ramp_mask, 1];  ramp_y = Y_pre[ramp_mask]
order       = np.argsort(ramp_x2);  ramp_x2, ramp_y = ramp_x2[order], ramp_y[order]
slope, intercept = np.polyfit(ramp_x2, ramp_y, 1)
ramp_fit    = intercept + slope * ramp_x2
ss_res      = np.sum((ramp_y - ramp_fit) ** 2);  ss_tot = np.sum((ramp_y - ramp_y.mean()) ** 2)
r2          = 1 - ss_res / ss_tot
print(f"\nCorner ramp (n = {ramp_mask.sum()}): y = {slope:.1f} * x2 + {intercept:.1f}   R^2 = {r2:.6f}   residual rmse = {np.sqrt(ss_res / len(ramp_y)):.4f}")
print("  pairwise slopes between neighbouring queries:", np.round(np.diff(ramp_y) / np.diff(ramp_x2), 1))
print("  -> the corner is a ramp that was still rising in x2; the Week 3 'cliff' is a single point at x2 = 0.796")

# 3. WEEK 13 CHANGE: which direction does my own data point to?
y_sl = signed_log1p(Y_pre)
print("\ncorr(signed-log y, x1..x4):", np.round([np.corrcoef(X_pre[:, k], y_sl)[0, 1] for k in range(4)], 2), " (x3, x4 positive; x2 positive along the ramp)")

# 4. WEEK 13 CHANGE: coverage of the 16 vertices of the unit hypercube, using a GP fitted on all 32 points
full_bounds  = np.array([[0.0, 0.999999]] * 4)
result_full  = bayesian_optimization_step(X_pre, Y_pre, full_bounds, beta = 1.0, kernel_type = "rbf")
vertices     = np.array(list(itertools.product([0.0, 0.999999], repeat = 4)))
mu_v, sd_v   = result_full["gpr"].predict(result_full["X_scaler"].transform(vertices), return_std = True)
mu_v_yield   = inverse_signed_log1p(mu_v * result_full["Y_std"] + result_full["Y_mean"])
sd_v_log     = sd_v * result_full["Y_std"]
near_dist    = np.array([np.min(np.linalg.norm(X_pre - v, axis = 1)) for v in vertices])
vert_label   = ["".join("1" if c > 0.5 else "0" for c in v) for v in vertices]
print("\nVertex coverage (label = x1x2x3x4, 1 means 0.999999):")
for k in np.argsort(-sd_v_log):
    print(f"  {vert_label[k]}  nearest observed point = {near_dist[k]:.3f}   GP mean = {mu_v_yield[k]:8.1f}   sigma(log) = {sd_v_log[k]:.2f}")
print("  -> most vertices are 0.5-0.8 from the nearest observation; 1111 is the farthest (0.78), and 13 of 16 share the same maximum uncertainty (sigma ~2.0): the GP cannot rank them.")


FINAL_MODE   = "basin_probe" # "basin_probe" = far vertex (hypothesis test); "ramp" = bounded step along the x2 ramp (x2 = 0.602)
if FINAL_MODE == "basin_probe":
    x_next   = np.array([0.999999, 0.999999, 0.999999, 0.999999])
    bounds   = full_bounds
    beta     = 1.0
    pred_mu  = float(mu_v_yield[vert_label.index("1111")])
    pred_std_log = float(sd_v_log[vert_label.index("1111")])
else:
    x_next   = np.array([0.0, 0.602000, 0.999999, 0.999999])
    bounds   = np.array([[0.0, 0.001], [0.57, 0.65], [0.998, 0.999999], [0.998, 0.999999]])
    beta     = 0.35
    pred_mu  = intercept + slope * x_next[1]
    pred_std_log = float('nan')

gpr             = result_full["gpr"]
X_scaler        = result_full["X_scaler"]
Y_mean          = result_full["Y_mean"]
Y_std           = result_full["Y_std"]
best_idx        = np.argmax(Y_pre)
best_x_now      = X_pre[best_idx]                            # still Week 10's point
ucb_next        = float('nan')                           # not an acquisition-function choice this week

print(f"\nWeek 13 (final) mode        : {FINAL_MODE}")
print(f"Week 13 (final) query       : {np.round(x_next, 6)}")
print(f"Model mean / sigma(log) : {pred_mu:.1f} / {pred_std_log:.2f}   (best before the query {Y_pre.max():.3f}; at the vertex the GP is effectively uninformed)")
````
